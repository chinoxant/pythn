 ## LISTS

Lists are one of the builtin python data structures they store multiple items in a single variable.
lists are;
1. Ordered
2. Mutable
3. Allow duplicates
Here is how we create a list:

```python
score = [45, 65, 76, 89, 90]
print("score:", score)
```

**ACCESSING LIST ELEMENTS**

Elements in a list can be accessed using their index. Python uses a zero based index, meaning the first element has an index of 0.

```python
print(score[0])
print(score[1])
print(score[2])
```

**MODIFYING LIST ELEMENTS**

One of the features of a list is mutability which means that elements can be changed after the list has been created

```python
score[0] = 100
print(score)
```

**LIST METHODS**

**append()**

The append() method adds a value to the end of the list

```python
score.append(91)
print(score)
```

**len()**

The len method is used to output the number of elememts in a list

```python
print(len(score))
```

**extend()**

The extend() method is used to add elements from another list to a list


```python
numbers = [41, 42, 43, 55]
numbers.extend(score)
print(numbers)
```

**remove()**

The remove method is used to remove elements in a list.

```python
numbers.remove(42)
print(numbers)
```

**clear()**

This is used to cleaar the elements in a list

```python
numbers.clear()
print(numbers)
```

**sort**

This is used to sort the elements in a list. This arranges the elements of a list in  ascending order .

```python
score.sort()
print(score)
```

**reverse()**

This reverses the order of the list

```python
score.reverse()
print(score)
```
